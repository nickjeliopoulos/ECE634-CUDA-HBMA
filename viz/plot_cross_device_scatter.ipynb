```python
import matplotlib.pyplot as plt
import numpy

plt.rcParams['legend.fontsize'] = 10
plt.rcParams.update({'font.size': 16})


# Configurable labels for each series
series_labels = ['Naive CPU', 'Optimized CPU', 'Optimized GPU', 'Ours']
series_colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']  # Blue, Orange, Green, Red

# Hardcoded data
block_sizes = [4, 8, 16, 32]
# Latencies for neighborhood search size 1
latencies_naive_n1 = [1533, 168, 38.9, 9.34]  
latencies_opcpu_n1 = [1348, 334, 29.5, 6.49]  
latencies_opgpu_n1 = [4696, 1029, 233, 26.9]
latencies_ours_n1  = [0.192, 0.209, 0.196, 0.115]

latencies_naive_n2 = [1571, 887, 87.2, 18.6]  
latencies_opcpu_n2 = [3361, 823, 63.8, 13.8]  
latencies_opgpu_n2 = [11858, 2784, 397, 59.4]
latencies_ours_n2  = [0.464, 0.203, 0.179, 0.168]

latencies_naive_n3 = [8230, 652, 147, 27.2]  
latencies_opcpu_n3 = [6241, 455, 106, 20.5]  
latencies_opgpu_n3 = [22201, 3260, 592, 92.6]
latencies_ours_n3  = [0.538, 0.266, 0.167, 0.201]

### Neighborhood search sizes
neighborhood_sizes = [1, 2, 3]

### Method latencies
latencies = [
	### Neighborhood 1
	[latencies_naive_n1, latencies_opcpu_n1, latencies_opgpu_n1, latencies_ours_n1],
	### Neighborhood 2
	[latencies_naive_n2, latencies_opcpu_n2, latencies_opgpu_n2, latencies_ours_n2],
	### Neighborhood 3
	[latencies_naive_n3, latencies_opcpu_n3, latencies_opgpu_n3, latencies_ours_n3]
]

### Create subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharey=True)

for i, ax in enumerate(axes):
	for series_idx, series in enumerate(series_labels):
		ax.plot(block_sizes, latencies[i][series_idx], color=series_colors[series_idx], linestyle='-', marker='o')
		ax.scatter(block_sizes, latencies[i][series_idx], color=series_colors[series_idx], label=series_labels[series_idx])
	ax.set_title(f'Neighborhood Search Size {neighborhood_sizes[i]}')
	ax.set_xlabel('Block Size')
	ax.set_xticks(block_sizes)
	ax.set_xticklabels(block_sizes)
	ax.set_ylabel('Latency (ms)')
	ax.set_yscale('log')
	ax.legend()
	ax.grid(False)

### Adjust layout
plt.tight_layout()
plt.show()
```
